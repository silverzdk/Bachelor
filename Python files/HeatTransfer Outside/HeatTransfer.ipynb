### Based on Handbook for Transversely Finned Tube Heat Exchanger Design chapter 2

Physical properties of the external heat-transfer medium, which is usually
air or flue gases, are determined at the average flow temperature.

## Calculation of heating surface

The heating surface on the outside is given as 

$$
A = A_r + A_t 
$$

Where $A_r$ is the surface of the fins and $A_t$ is the surface of the tube minus the area occupied by the fins.

For circular fins the area is given as
$$
A_r = \frac{π}{2}( D^2 - d^2 + 2D \delta_r) \frac{L_{rs}}{s_r} z 
$$

And for rectangular fins

$$
A_r = 2 (c_{sq}^2 - 0.785 d^2 + 2 c_{sq}\delta_r) \frac{L_{rs}}{s_r} z 
$$

For the above equations the following nomenclature is used 

$D$ is the diameter of the fin, $d$ is the outside diameter of the tube, $\delta_r$ is the fin thickness, $L_{rs}$ length of the finned segment of the tube, $s_r$ fin spacing, $z$ total number of tubes in the tube bundle, $c_{sq}$ side length of the square fin.

## Overall heat transfer coefficient

The overall heattransfer coefficient is given as 

### $$ U = \frac{Ψ}{\frac{A}{A_in} \frac{1}{h_2} + \frac{A}{A_in}R_t + \frac{1}{h_{1rdc}}}$$

Where $\Psi$ is the thermal efficency factor, $A$ as above, $A_{in}$ is the area inside the tubes, $h_2$ is the heat transfer coefficent between the internal wall to internal medium, $R_t$ is the thermal resistance of the tube, $h_{1rdc}$ is a reduced heat transfer coefficent for the outside gas. 

### Convective heat transfer coefficient $h_c$

To determine the reduced heat transfer coefficient the convective heat transfer coefficient must first be determined. 


##### Gas velocity
The gas velocity is given as

$$
u_g = \frac{\dot m_g}{F ρ_g}
$$

Where $F$ is the minimum area, for staggered bundles the following parameter is used 
$$
\varphi_{cl} = \frac{S_1 - d_{cl}}{S_2' - d_{cl}}
$$

Where 
$$
d_{cl} = d + \frac{2 l_r \delta_r}{s_r}
$$

For inline and staggered tube bundles with $\varphi_{cl}\le 2$ the minimum area is given as

$$
F = a b - z_1 L_{c\cdot cr \cdot s}d_{cl} 
$$

for staggered tube bundles with $\varphi_{cl} > 2$
$$
F = (a b-z_1 L_{c\cdot cr \cdot s}d_{cl} ) \frac{2}{\varphi_{cl}}
$$

Where a and b is the height and width of a the duct, $L_{c\cdot cr \cdot s}$ is the length of the tube projected unto the cross section 

##### Convective coefficient

$$
h_c = 1.13 C_z C_q \frac{k_g}{d} \left( \frac{u_g d}{\nu_g} \right)^n \text{Pr}_g^{\frac{1}{3}}
$$

Where

$$
n = 0.7 + 0.08 \tanh(X) + 0.005 \Psi_r
$$
$$
C_q = (1.36 - \tanh(X))\left( \frac{1.1}{\Psi_r +8}-0.014 \right) 
$$

For staggered
$$
(X)_{staggered} = \frac{\sigma_1}{\sigma_2} - \frac{1.26}{\Psi_r}-2
$$
For inline
$$
(X)_{inline} = 4 \left( 2 + \frac{\Psi_r}{7}-\sigma_2 \right) 
$$

The fin coefficient for circular fins is given as
$$
(\Psi_r)_{circular} = \frac{1}{2d s_r}(D^2 -d^2 + 2 D\delta_2) +1 - \frac{\delta_1}{s_r}
$$
and for square fins
$$
(\Psi_r)_{square} = \frac{2 \left( c_{sq}^2 - 0.785 d^2 + 2c_{sq}\delta_A \right)  }{\pi d s_r} + 1 - \frac{\delta_r}{s_r}
$$

If there are more than 8 rows of tubes in the transverse direction ($z_2\ge8$) we find that
$$
C_z = 1
$$

otherwise

For inline and for staggered with $\sigma_1/\sigma_2 \ge 2$
$$
C_z = 3.5 z_2 ^{0.03} - 2.72
$$

For staggered with $\sigma_1/\sigma_2 \le 2$
$$
C_z = 3.15 z_2 ^{0.05} - 2.50
$$


### Reduced convective heat transfer coefficient $h_{1rdc}$

The reduced convective heat transfer coefficient is given as 
$$
h_{1rdc} = \left( \frac{A_r}{A} E \mu_r \psi_E + \frac{A_t}{A}  \right) h_c 
$$

Where $E$ is the fin efficiency. If a fin efficiency for circular fins is used the area of a circular fin must also be used. 

To account for nonuniformity of the heat transfer over the fin surface the following factor is calculated.
$$
\psi_E = 1 - 0.016 \left( \frac{D}{d}-1 \right) (1+\tanh(2 \beta l_r -1))
$$

The coefficent $\mu_r$ is to account for fins of non constant thickness, something that is not looked at here.

```python
Geom = Geometry()
import numpy as np

Geom.Tube.set_geometry(
    wall_thickness=2.5e-3,
    outer_diameter=0.028,
)

Geom.Duct.set_geometry(
    a=0.56,
    b=0.5,
)

Geom.Fin.set_geometry(
    average_fin_thickness = 8e-4,
    fin_height = 0.0135,
    fin_spacing = 0.003,
)

Geom.Bank.set_geometry(
    arrangement = arrangementType.Staggered,
    transverse_pitch = 0.059,
    longitudinal_pitch = 0.059 * np.sqrt(3)/2,
    number_of_rows = 9,
)

def min_area():
    # Calculate equivalent diameter
    d_cl = Geom.Tube.outer_diameter + (2* Geom.Fin.fin_height*Geom.Fin.average_fin_thickness) / Geom.Fin.fin_spacing
    
    S_1 = Geom.Bank.transverse_pitch
    S_2 = Geom.Bank.longitudinal_pitch
    
    if Geom.Bank.arrangement == arrangementType.Inline:
        phi_cl = 1
    elif Geom.Bank.arrangement == arrangementType.Staggered:
        phi_cl = (Geom.Bank.transverse_pitch-d_cl)/(Geom.Bank.diagonal_pitch-d_cl)
    
    
    if Geom.Bank.arrangement == arrangementType.Inline or phi_cl <= 2:
        F = Geom.Duct.area - Geom.Bank.number_of_rows * d_cl * Geom.Duct.b
    else:
        F = (Geom.Duct.area - Geom.Bank.number_of_rows * d_cl * Geom.Duct.b) * 2/ phi_cl
    
    return F

min_area()
```
